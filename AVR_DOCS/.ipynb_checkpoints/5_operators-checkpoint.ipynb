## The join() method in Python is used to concatenate a list of strings with a specified delimiter. 

For example, if you have a list of strings words = ['hi', 'hello', 'bonjour'] and you want to join them into a single string with a space as the delimiter, you can use ' '.join(words). This will return the string 'hi hello bonjour'.

The join() method is called on the delimiter string and is passed the list of strings as an argument. The resulting string will have the delimiter string between each element of the list and at the end of the last element.

It can also be used on a tuple and it will give the same results.

It's important to note that the join method will return a new string and it will not modify the original list.

```python
words = ["hi" , "hello" , "bonjour"]
```

```python
new_string = ' '.join(words)
```

```python
new_string
```

```python
words
```
