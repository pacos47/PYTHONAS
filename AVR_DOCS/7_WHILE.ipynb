```python
k = 0;
while k >= 0:
    print('μπλοκ-εντολών-1',k)
    k +=1
    if k<7 :
        continue # πήγαινε στην αρχή
    if 5 < 10 :
        print('breaking the loop...')
        break # βγες από το βρόχο
else:
    print('μπλοκ-εντολών-2 # αν τέλειωσε χωρίς break')
print('out of while...')
```
