## next example 
## In Python, when working with lists, the slice notation a[:] refers to the entire list.
### It creates a new copy of the list and creates a new reference to the new copy, so any change made to the new copy will not affect the original list.

### For example, if you have the list a = [1, 2, 3], then a[:] would return a new list [1, 2, 3] that is an exact copy of the original list.
### You can also use this notation to create a sublist by providing the start and end index. For example, a[1:3] returns [2,3] which is sublist from index 1 to 2(not including 3).

It is useful when you want to use a copy of the list instead of modifying the original one, or when you want to create a sub-list.

```python
a = [1 , 2 ,3]
```

```python
b = a[:]
```

```python
b
```

```python
a[0] = 8
```

```python
a
```

```python
b
```
