## dir(object)
Definition and Usage
The dir() function returns all properties and methods of the specified object, without the values.

This function will return all the properties and methods, even built-in properties which are default for all object.

```python
## Display the content of an object:

class Person:
  name = "John"
  age = 36
  country = "Norway"

print(dir(Person))
```

```python
help(Person)
```


### FORMATING
In Python 3, there are several ways to format strings, including:

The % operator: This uses the % symbol to replace placeholders in a string with values.

For example, "Hello, %s" % name would replace %s with the value of the variable "name".

#### The format() method: This method uses {} placeholders that are replaced with values passed to the method. 

For example, 
#### "Hello, {}".format(name) would replace {} with the value of the variable "name".

## f-strings: This is a relatively new way of formatting strings available in Python 3.6 and later. It allows you to embed expressions inside string literals using {}.

## For example, f"Hello, {name}" would replace {} with the value of the variable "name".

Formatter: This is an advanced method that allows you to create complex string formatting templates. It is useful for creating reusable templates.

Template Strings: This method is similar to the format() method, but it uses $ sign instead of {}

String Interpolation : This method uses the str.format() and f-strings method to create a string with the formatted values.




# ακριβεια αριθμών

### '{:.25f}'.format(q) is a Python string formatting expression that formats the floating point number 'q' to have 25 decimal places. The 'f' in the expression specifies that 'q' is a floating point number, and the '.25' before the 'f' specifies that the number should be formatted with 25 decimal places. The entire expression is enclosed in curly braces {} and is used in conjunction with the string format() method.

```python
q = 0.3;
'{}'.format(q)
```

```python
'{:.25f}'.format(q)
```

```python
f"katitis,{q}"
```
