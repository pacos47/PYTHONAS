```python
grade = 4;
if grade >= 5 :
        result = 'pass'
else:
        result = 'fail'
```

# sorted() function  => Syntax : new_item = sorted(item) so preserves item
### The sorted() function in Python is used to sort a list, tuple, or other iterable object. It returns a new sorted list from the elements of the iterable. By default, it sorts the elements in ascending order.

For example,

```python
numbers = [3, 1, 4, 2]
sorted(numbers)
```

```python
print(numbers)
```

```python
sorted_numbers = sorted(numbers)
print(sorted_numbers)
```
