## In Python, the f-string (also known as "formatted string literals") is a way to embed expressions inside string literals. This feature was introduced in Python 3.6 and provides a concise and readable way to create formatted strings.

## To create an f-string, you simply prefix the string literal with the letter 'f' (or 'F') and then include expressions inside curly braces {}.

# Minimals : f""  , it does nothing  , f"some string" : prints some string , f"foo {bar}" prints : foo bar's value



```python
name = "Alice"
age = 30
print(f"My name is {name} and I'm {age} years old.")
```

## As you can see, the expressions inside the curly braces are evaluated at runtime and their values are inserted into the string. You can include any valid Python expression inside the curly braces, including variables, literals, function calls, and arithmetic expressions.

You can also use 
# format specifiers 
to control the formatting of the values. For example:


```python
pi = 3.141592653589793
print(f"The value of pi is approximately {pi:.3f}")
```

In this example, the format specifier 
# :.3f 
is used to format the value of pi as a floating-point number with 3 decimal places.

# "foo {}".format(bar)
In Python, "foo {}".format(bar) is a string formatting expression that creates a string by combining a fixed string "foo " with a value of the bar variable.

The curly braces {} are placeholders that represent where the value of the bar variable should be inserted. The .format() method is called on the string "foo {}", passing the bar variable as an argument, which is then inserted into the string at the appropriate placeholder location.

Here's an example:

```python
bar = "world"
s = "Hello, {}!".format(bar)
print(s)
```

### In this example, the value of the bar variable is the string "world". The .format() method is called on the string "Hello, {}!", passing the bar variable as an argument. The resulting string, "Hello, world!", is then assigned to the variable s.

### You can also include multiple placeholders in the same string and pass multiple arguments to the .format() method to replace each placeholder with a corresponding value. For example:

```python
x = 10
y = 20
z = x + y
s = "{} + {} = {}".format(x, y, z)
print(s)
```

### In this example, the string "{} + {} = {}" contains three placeholders that are replaced with the values of x, y, and z, respectively. The resulting string, "10 + 20 = 30", is then assigned to the variable s.
