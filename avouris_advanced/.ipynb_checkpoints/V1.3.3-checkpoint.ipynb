### άσκηση : ταξινόμηση ενός λεξικού με ανώνυμη συνάρτηση
Με αντίστοιχο τρόπο που είδαμε την ταξινόμηση αντικειμένων μπορεί να γίνει ταξινόμηση στοιχείων ενός λεξικού, όπως το παρακάτω:

dd = {1:{'a':8,'b':10}, 8:{'a':2,'b':6}, 3:{'a':7,'b':1}}

### dict of dict
Ζητείται να ταξινομηθεί ως προς το κλειδί ‘b’
# dict.items() = list of tuples


```python
dd = {1:{'a':8,'b':10}, 8:{'a':2,'b':6}, 3:{'a':7,'b':1}}
print(sorted(dd.items()))   ## sorted based on key (default)

for i in sorted(dd.items(), key=lambda x: x[1]['b']):  ## [second tuple item ie dict][index within dict]
       print(i)
```

In Python, you can 
## add a new key-value pair to a dictionary using the square bracket notation or the .update() method. 
Here are examples of both methods:
# Using square bracket notation
my_dict = {"apple": 2, "banana": 4}
my_dict["orange"] = 6
print(my_dict)   # Output: {"apple": 2, "banana": 4, "orange": 6}

# Using the .update() method
my_dict = {"apple": 2, "banana": 4}
my_dict.update({"orange": 6})
print(my_dict)   # Output: {"apple": 2, "banana": 4, "orange": 6}


```python
# employee example
class Employee():
    ''' Ο εργαζόμενος σε μια επιχείρηση '''
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

# main program
the_employees = [] # just a list
while True:
    name = input('Όνομα:')
    if not name: break
    salary = input('Μισθός:')
    the_employees.append(Employee(name, salary))  # now a list of objects

# print the employees
print('\nΟι υπάλληλοι είναι:')
for employee in sorted(the_employees, key=lambda x: x.name):
    print(employee.name, employee.salary, sep='\t')
```

### sort dictionary , no objects

```python


# main program
the_employees = {}
while True:
    name = input('Όνομα:')
    if not name: break
    salary = input('Μισθός:')
    the_employees[name] = salary

# print the employees
print('\nΟι υπάλληλοι είναι:')
print(the_employees)
for i in sorted(the_employees.items(), key=lambda x: x[0]):  
       print(i)
```

```python
li = [7, 2, 5]
print(sorted(li, key=lambda x: -x)) # μου γυρνας για ΓΝΩΡΙΣΜΑ το -7  που ειναι το μικροτερο απο τα -7,-2,-5 => ταξινομηση 7,5,2
```
