# __str__
### __str__ is a special method in Python that returns a string representation of an object.
### It is often used to provide a human-readable description of an object's state or value. When you call the print() function on an object, it implicitly calls the object's __str__ method to get a string representation of the object.

For example, consider a simple class that represents a book:



```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"{self.title} by {self.author}"
```

### In this class, the __str__ method returns a string representation of the book object that includes its title and author. If we create an instance of this class and call print() on it, we'll see the string representation that is returned by the __str__ method:



```python
book = Book("The Catcher in the Rye", "J.D. Salinger")
print(book)
```

### Note that the __str__ method should always return a string. If it returns anything else, Python will raise a TypeError.
