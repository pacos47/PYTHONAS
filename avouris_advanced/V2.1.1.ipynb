In programming, a namespace is a container that holds a collection of identifiers (e.g., variable names, function names, class names) and their corresponding objects (e.g., values, functions, classes). A namespace provides a way to organize the identifiers in a program and to avoid naming conflicts between identifiers that are defined in different parts of the program.

In Python, a namespace is implemented as a dictionary that maps the identifiers to their corresponding objects. There are several types of namespaces in Python, including the built-in namespace, the global namespace, and the local namespace.

The built-in namespace is a namespace that contains all the built-in functions, classes, and variables in Python. These identifiers are available to all Python programs without the need for importing any modules.

The global namespace is a namespace that contains all the identifiers that are defined in the global scope of a Python program (i.e., outside of any functions or classes). These identifiers are accessible from anywhere in the program.

The local namespace is a namespace that contains all the identifiers that are defined within a function or a method. These identifiers are accessible only within the function or method where they are defined.

When Python encounters an identifier in a program, it first searches for the identifier in the local namespace of the current function or method. If the identifier is not found in the local namespace, Python then searches for the identifier in the global namespace. If the identifier is still not found in the global namespace, Python finally searches for the identifier in the built-in namespace.

If two or more identifiers have the same name in the same namespace, this is called a naming conflict. To avoid naming conflicts, you can use different namespaces or qualify the identifiers with their namespace (e.g., module1.variable1, module2.variable1).







In Python, __main__ is the name of the namespace that is used for the script that is currently being executed. When you run a Python script, Python creates a new namespace with the name __main__, and all the variables, functions, and classes defined in the script are added to this namespace.

The __main__ namespace is used to differentiate the variables, functions, and classes defined in the script being executed from those defined in imported modules. When a module is imported, it gets its own namespace, which is separate from the __main__ namespace.

Here's an example:

Suppose you have a Python script named my_script.py with the following code:

```python
# my_script.py

def my_function():
    print("Hello, world!")

class MyClass:
    def __init__(self, value):
        self.value = value

if __name__ == "__main__":
    # This code is executed only if this script is run as the main program
    x = MyClass(42)
    print(x)
    print(x.value)
    my_function()
```

When you run this script using the command python my_script.py, Python creates a new namespace with the name __main__, and adds the variables, functions, and classes defined in the script to this namespace. The if __name__ == "__main__": block is used to check if this script is being run as the main program. If so, it executes the code in the block.

The __name__ variable is a built-in variable in Python that contains the name of the current module or script. When a Python script is executed, the value of __name__ is set to "__main__" to indicate that this script is being run as the main program. This allows you to include code that should only be executed when the script is run as the main program (as opposed to being imported as a module).

__str__ is a special method in Python that is used to define the string representation of an object. When you print an object, Python uses the __str__ method to convert the object to a string that can be displayed on the screen or stored in a file.

The __str__ method should return a string that represents the object in a human-readable format. By default, the __str__ method returns a string that includes the name and memory address of the object (e.g., <__main__.MyClass object at 0x7f07d437ebe0>). However, you can override the __str__ method to return a custom string representation of the object.

Here's an example:



```python
class MyClass:
    def __init__(self, value):
        self.value = value

    def __str__(self):
        return f"MyClass({self.value})"

x = MyClass(42)
print(x)
```

## Πώς θα επιτρέπαμε στην κλάση Employee να γνωρίζει τη λίστα των εργαζόμενων;
Θα μεταφέρουμε τη λίστα των εργαζομένων the_employees μέσα στην κλάση, ως γνώρισμα κλάσης (class attribute). Τα γνωρίσματα αυτά ορίζονται στο επίπεδο της κλάσης και είναι γνωστά σε όλα τα αντικείμενα. Για παράδειγμα η λίστα αυτή θα πρέπει να ενημερώνεται από το δημιουργό αντικειμένων για κάθε νέο αντικείμενο.

Εν γένει γίνεται αναφορά στα γνωρίσματα κλάσης με σημειογραφία <ΌνομαΚλάσης>.<ΌνομαΓνωρίσματος>

```python
#νέα έκδοση της κλάσης Employee
# employee example
class Employee():
    ''' Ο εργαζόμενος σε μια επιχείρηση '''
    the_employees = []
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.the_employees.append(self)    ##κλειδί το self : οταν δημιοyργειται ενα αντικειμενο(self)
                                                ## προστίθεται στην λιστα : the_employees (γνώρισμα τησ ταξης Emmployee)

# main program

while True:
    name = input('Όνομα:')
    if not name: break   ## με το πατημα ενος χαρακτηρα το name =True , so just type enter to be False and break
    salary = input('Μισθός:')
    Employee(name, salary)

# print the employees
print('\nΟι υπάλληλοι είναι:')
for employee in sorted(Employee.the_employees, key=lambda x: x.name):
    print(employee.name, employee.salary, sep='\t')
```
