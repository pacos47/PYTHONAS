```python
from math import sqrt

class Point:  # capitalize class names
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def distance(self, a, b):  # take a and b as arguments, not inputs
        dist = sqrt((self.x - a)**2 + (self.y - b)**2)
        return dist
    
p = Point(1, 1)

print(p.distance(3, 3))  # pass arguments to the method
```

Added self as the first argument to the distance method so that it can access the x and y attributes of the object.
Added a print statement to display the result of calling the distance method with arguments (2, 2).
