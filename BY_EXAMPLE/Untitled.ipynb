## functions with unknown number of parameters
## *foo , foo=list , **bar , bar=dictionary

```python
def fonc(*xyma , **pairs) :
    for a in xyma : print(a)
    for k,v in pairs.items() : print(k,v)
```

```python
fonc('kiki','koko','klp' , miky='mouse' , donald = 17)
```

```python
def f(**x):
	print(x.values())
f(a=8,b=12,c=18)
```

```python
def f(**x):
	return [str(x) for x in x.values()]
print('-'.join(f(a=8,b=12,c=18)))
```
