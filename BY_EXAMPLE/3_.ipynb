### In Python, the terms "object method" and "object function" are often used interchangeably, but there is a subtle difference between the two.

An object method is a function that is associated with an object and is called on that object. For example, the upper method in the previous example (see CLASSS_INSTANCE_METHOD.ipynb) is a 
## string object method, and it's called on a string object. 
Object methods 
# operate on the data contained within an object and can modify that data OR return a new value based on the data.

# On the other hand, an object function is a standalone function that takes an object as an argument and operates on that object. For example,
## the len function in Python is a built-in object function, 

## which returns the number of items in an object. Object functions 

## do not modify the object passed as an argument and instead return a new value based on the object.

In general, the distinction between object methods and object functions is not critical and the terms are often used interchangeably. However, it's useful to understand the subtle difference between them to better understand the behavior and use of different functions and methods in Python.

moves it to the end of the word and adds on an
“ay”. If a word begins with a vowel you just add
“way” to the end. For example, pig becomes igpay,
banana becomes ananabay, and aadvark becomes
aadvarkway. Create a program that will ask the
user to enter a word and change it into Pig Latin.
Make sure the new word is displayed in lower case.

### The statement print(f" pi = {pi:.2f}") in Python is using 

# f-strings (formatted string literals).

### The f before the string indicates that this is an f-string and the expressions inside the curly braces {} are evaluated and replaced with their values.
### The expression inside the braces {pi:.2f} specifies the value of the variable pi (which should be a float value) and the formatting that should be applied to it.
### The .2f part specifies the format of the float value. The .2 specifies that only 2 digits should be after the decimal point, and the f specifies that the value should be formatted as a float.
### So, when the statement is executed, the value of pi is inserted into the string in the place of {pi:.2f} and the resulting string is printed to the console.

For example, if pi = 3.14159265359, then the statement print(f" pi = {pi:.2f}") will print the following output to the console:

```python
print(f"{pi:.2f}")           # f-string
```
