## A turtle will travel along a path that you define, leaving a pen mark behind it. As you control
the turtle, the pattern that is left is revealed. To draw the pentagon shown below you would
type in the following code.


```python
import turtle
turtle.shape("turtle")
for i in range(5) :
    turtle.forward(100)
    turtle.right(72)
    
turtle.exitonclick()
```

```python
import turtle
t = turtle.Turtle()

for i in range(4):
    t.forward(100)
    t.right(90)

turtle.done()
```

### In the previous example, t is a variable that is used to refer to an instance of the turtle.Turtle class.
### The line t = turtle.Turtle() creates an instance of the Turtle class and assigns it to the variable t.
#### me : the turtle(small t) in the same line stand for : use the turtle library , like math.pi indicatea the pi of the math module
### This allows you to interact with the turtle object and use its methods, such as forward and right, to control the turtle's movement and drawing. In this way, t acts as a handle to the turtle, and you can use it to manipulate the turtle on the screen.
