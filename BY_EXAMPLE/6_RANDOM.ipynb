```python
tosses = 10000000
count_h = 0
count_t = 0

while tosses != 0 :
    face = random.choice(['head', 'tail'])
    tosses -=1
    if face == 'head' : count_h +=1
    else : count_t +=1
print(count_h/count_t)
    
```
