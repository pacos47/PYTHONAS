# The expression 'still there?'.upper() calls the upper method on the string 'still there?', which returns a new string with all the characters of the original string converted to uppercase. The result of this expression is 'STILL THERE?'.

When you assign this result to the variable bar, the following line of code is executed:

python
Copy code
bar = 'STILL THERE?'
So, after this line of code 

```python
bar = 'still there?'.upper()
```

```python
bar
```

# The expression str.lower("fOo") calls the lower METHOD on the str class, and passes the string "fOo" as an argument. The lower method returns a new string with all the characters of the original string converted to lowercase. In this case, the result of the expression str.lower("fOo") would be "foo".

```python
foo = str.lower(bar)
```

```python
foo
```
