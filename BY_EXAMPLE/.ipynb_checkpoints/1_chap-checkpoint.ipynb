#### To display a more 
### human-friendly number, you can round or format the output:

#### Using round():

python

pounds = 5.7 * 2.20462

print(f"There are {round(pounds, 2)} pounds in 5.7 kilos")

##### Output: There are 12.56 pounds in 5.7 kilos

#### Using string formatting:

python

print(f"There are {pounds:.2f} pounds in 5.7 kilos")

##### Output: There are 12.56 pounds in 5.7 kilos
